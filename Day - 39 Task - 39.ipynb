```python
import faiss
import numpy as np

# Sample documents
documents = [
    "Python is used for Artificial Intelligence",
    "FAISS performs vector similarity search",
    "Machine Learning learns from data",
    "Semantic search understands meaning",
    "Vector databases store embeddings",
]

# Manual embeddings (example vectors)
embeddings = np.array([
    [0.9, 0.1, 0.2],
    [0.8, 0.2, 0.1],
    [0.2, 0.9, 0.8],
    [0.85, 0.15, 0.1],
    [0.75, 0.25, 0.2]
], dtype=np.float32)

# Create FAISS index
dimension = embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
index.add(embeddings)

print("Indexed", len(documents), "documents")

# Query embedding
query_embedding = np.array([[0.82, 0.18, 0.12]], dtype=np.float32)

# Search top 3
k = 3
distances, indices = index.search(query_embedding, k)

print("\nSemantic Search Results")
print("=" * 40)

for rank, idx in enumerate(indices[0], start=1):
    print(f"\nRank #{rank}")
    print("Document:", documents[idx])
    print("Distance:", round(float(distances[0][rank-1]), 4))
```
